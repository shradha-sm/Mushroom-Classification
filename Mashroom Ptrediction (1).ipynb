```python
# Step 6: Data Preprocessing.
```

```python
# Check Unique Values in Each Column

for col in df.columns:
    print(f"{col}: {df[col].unique()}")
```

```python
# Drop veil-type column

df.drop('veil-type', axis=1, inplace=True)
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

for col in df.columns:
    df[col] = le.fit_transform(df[col])
```

```python
df.head()
```

```python
# Correlation Matrix
correlation = df.corr()

# Plot Heatmap
plt.figure(figsize=(15, 12))
sns.heatmap(correlation, annot=False, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

```python
# Step 7: Feature Selection.
```

```python
# Separate Features and Target

X = df.drop('class', axis=1)
y = df['class']
```

```python
print("Features Shape:", X.shape)
print("Target Shape:", y.shape)
```

```python
# Step 8: Train-Test Split.
```

```python
from sklearn.model_selection import train_test_split
```

```python
# Split Dataset into Train and Test

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
# Step 9: Model Building.
```

```python
# Logistic Regression.
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
# Create Logistic Regression Model

lr = LogisticRegression(max_iter=1000, random_state=42)
```

```python
# Train the Model

lr.fit(X_train, y_train)
```

```python
# Predict on Test Data

y_pred_lr = lr.predict(X_test)
```

```python
# Accuracy Score

accuracy = accuracy_score(y_test, y_pred_lr)
print("Logistic Regression Accuracy:", accuracy)
```

```python
# Confusion Matrix

cm = confusion_matrix(y_test, y_pred_lr)
print(cm)
```

```python
# Classification Report

print(classification_report(y_test, y_pred_lr))
```

```python
# Observation:
# Logistic Regression achieved an accuracy of 95%.
# The model performed well for both edible and poisonous mushrooms.
# Precision, Recall, and F1-Score are approximately 95%, indicating good classification performance.
```

```python
# Dicision Tree.
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dt = DecisionTreeClassifier(random_state=42)
```

```python
dt.fit(X_train, y_train)
```

```python
y_pred_dt = dt.predict(X_test)
```

```python
print("Decision Tree Accuracy:", accuracy_score(y_test, y_pred_dt))
```

```python
print(confusion_matrix(y_test, y_pred_dt))
```

```python
print(classification_report(y_test, y_pred_dt))
```

```python
# Observation:
# Decision Tree Classifier achieved 100% accuracy on the test dataset.
# The model correctly classified all edible and poisonous mushrooms.
# Precision, Recall, and F1-Score are all equal to 1.00, indicating excellent performance.
```

```python
# Random Forest.
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)
```

```python
rf.fit(X_train, y_train)
```

```python
y_pred_rf = rf.predict(X_test)
```

```python
print("Random Forest Accuracy:", accuracy_score(y_test, y_pred_rf))
```

```python
print(confusion_matrix(y_test, y_pred_rf))
```

```python
print(classification_report(y_test, y_pred_rf))
```

```python
# K-Nearest Neighbors (KNN)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn = KNeighborsClassifier(n_neighbors=5)
```

```python
knn.fit(X_train, y_train)
```

```python
y_pred_knn = knn.predict(X_test)
```

```python
print("KNN Accuracy:", accuracy_score(y_test, y_pred_knn))
```

```python
print(confusion_matrix(y_test, y_pred_knn))
```

```python
print(classification_report(y_test, y_pred_knn))
```

```python
# Observation:
# K-Nearest Neighbors (KNN) achieved approximately 100% accuracy.
# The model classified almost all mushrooms correctly.
# Precision, Recall, and F1-Score indicate excellent classification performance.
```

```python
# Step 10: Model Comparison.
```

```python
# Compare Model Accuracy

results = {
    'Model': ['Logistic Regression', 'Decision Tree', 'Random Forest', 'KNN'],
    'Accuracy': [
        accuracy_score(y_test, y_pred_lr),
        accuracy_score(y_test, y_pred_dt),
        accuracy_score(y_test, y_pred_rf),
        accuracy_score(y_test, y_pred_knn)
    ]
}

import pandas as pd

comparison = pd.DataFrame(results)
comparison
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x='Model', y='Accuracy', data=comparison)

plt.title("Model Comparison")
plt.ylim(0.90, 1.01)

plt.show()
```

```python
# Observation:
# Decision Tree and Random Forest achieved the highest accuracy (100%).
# KNN also performed exceptionally well.
# Logistic Regression achieved 95% accuracy.
# Decision Tree and Random Forest are the best-performing models for this dataset.
```
